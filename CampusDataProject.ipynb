Convert recorder ids to their numbers for simplification

```python
df['recorder_id'] = pd.to_numeric(
    df['recorder_id'].astype(str).str.extract(r'(\d+)$')[0],
    errors='coerce'
).fillna(-1).astype(int).astype(str)

df['recorder_id'] = df['recorder_id'].replace('-1', 'Unknown')

display(df.head(134))
```
